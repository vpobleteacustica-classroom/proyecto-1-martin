# Clasificador de Música por Género
### Primer Hito de Entrega  
Integrantes: Martin Arigo

## Objetivo General
**Desarrollar un sistema que reciba una canción y prediga su género musical a partir de características acústicas, usando Python y datasets públicos.**



## Objetivos Específicos

1. Investigar y cargar datasets de música (por ejemplo, GTZAN y MusicGenreClassification).
2. Procesar los audios y extraer características relevantes con librerías de Python (librosa, etc.).
3. Entrenar un clasificador básico (KNN/SVM) que asigne géneros musicales a canciones nuevas.
4. Evaluar el rendimiento del clasificador con ejemplos de prueba y métricas
5. Documentar y presentar resultados comparando con referencias del estado del arte.


## Estado del Arte
Todo empezó con Tzanetakis y Cook (2002), que fueron los pioneros en clasificar música por computadora. Usando características del sonido y algoritmos simples como KNN, lograron un 61% de acierto con 10 géneros.

Hoy en día, los métodos más modernos usan redes neuronales profundas que son muy precisas, pero tienen un problema, es difícil entender por qué toman sus decisiones. Un ejemplo de este enfoque es el trabajo de Lachmish (2021), quien implementó un sistema de clasificación utilizando redes neuronales convolucionales.

El enfoque de este trabajo consistirá en volver a lo básico pero con herramientas mejores:

Usará librosa (herramienta moderna) para analizar la música

Probará métodos clásicos como KNN, SVM y Random Forest

## 	Materiales y Métodos

**Materiales / Datos**
- Datasets: GTZAN data set de 10 generos de musica con 1000 canciones, MusicGenreClassification 
- Librerías Python: librosa, numpy, pandas, scikit-learn, matplotlib.

**Metodologia**

1. Carga y preprocesamiento: Cargar los archivos de audio y normalizarlos para garantizar una escala de amplitud uniforme.

2. Extracción de características: Utilizar librosa para extraer un conjunto de características acústicas relevantes (MFCCs, Chroma, Spectral Contrast, Tempo, Zero-Crossing Rate).

3. Construcción del dataset: Organizar las características extraídas y sus etiquetas correspondientes en un DataFrame estructurado.

4. División de datos: Separar el dataset en conjuntos de entrenamiento y prueba para evaluar el rendimiento del modelo.

5. Entrenamiento de modelos: Entrenar y ajustar los hiperparámetros de clasificadores clásicos (KNN, SVM, Random Forest).

6. Evaluación y análisis: Medir el rendimiento de los modelos utilizando métricas como precisión (accuracy), exhaustividad (recall) y puntuación F1(equilibrio del modelo), presentando los resultados de forma comparativa.
