# Clasificador de Música por Género
### Primer Hito de Entrega  
Integrantes: Martin Arrigo

## Objetivo General
**Desarrollar un sistema que reciba una canción y prediga su género musical a partir de características acústicas, usando Python y datasets públicos.**



## Objetivos Específicos

1. Investigar y cargar datasets de música (por ejemplo, GTZAN y MusicGenreClassification).
2. Procesar los audios y extraer características relevantes con librerías de Python (librosa, etc.).
3. Entrenar un clasificador básico (KNN/SVM) que asigne géneros musicales a canciones nuevas.
4. Evaluar el rendimiento del clasificador con ejemplos de prueba y métricas
5. Documentar y presentar resultados comparando con referencias del estado del arte.


## Estado del Arte

Tzanetakis y Cook (2002) establecieron el estándar para clasificación de géneros musicales usando características acústicas y algoritmos tradicionales como KNN y SVM. Lograron 61% de precisión en 10 géneros con el dataset GTZAN.

Nuestro proyecto aplicará este mismo enfoque usando herramientas modernas como `librosa` para extraer características y clasificadores tradicionales para la predicción.



## 	Materiales y Métodos

**Materiales / Datos**
- Datasets: GTZAN, MusicGenreClassification (o equivalentes disponibles públicamente).
- Librerías Python: librosa, numpy, pandas, scikit-learn, matplotlib.

1. Cargar y normalizar los archivos de audio.
2. Extraer características acústicas (MFCCs, Chroma, Spectral Contrast, Tempo, Zero-Crossing Rate).
3. Construir una tabla (DataFrame) con características y etiquetas (género).
4. Dividir en conjuntos de entrenamiento y prueba.
5. Entrenar clasificadores (KNN, SVM como referencia).
6. Evaluar (accuracy, precision, recall, F1) y presentar resultados.
